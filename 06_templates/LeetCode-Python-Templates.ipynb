## 1. Hash Map — Two Sum (LC 1)
**When:** complement lookup, seen-before

````python
# SOLUTION
def two_sum(nums: list[int], target: int) -> list[int]:
    seen = {}
    for i, num in enumerate(nums):
        need = target - num
        if need in seen:
            return [seen[need], i]
        seen[num] = i
    return []
````

## 2. Two Pointers — Valid Palindrome (LC 125)

````python
# SOLUTION
def is_palindrome(s: str) -> bool:
    l, r = 0, len(s) - 1
    while l < r:
        while l < r and not s[l].isalnum(): l += 1
        while l < r and not s[r].isalnum(): r -= 1
        if s[l].lower() != s[r].lower(): return False
        l += 1; r -= 1
    return True
````

## 3. Sliding Window — Longest Substring (LC 3)

````python
# SOLUTION
def length_of_longest_substring(s: str) -> int:
    seen = {}
    left = best = 0
    for right, ch in enumerate(s):
        if ch in seen and seen[ch] >= left:
            left = seen[ch] + 1
        seen[ch] = right
        best = max(best, right - left + 1)
    return best
````

## 4. Prefix Sum — Subarray Sum Equals K (LC 560)
**Critical:** initialize `freq = {0: 1}`

````python
# SOLUTION
def subarray_sum(nums: list[int], k: int) -> int:
    count = prefix = 0
    freq = {0: 1}
    for num in nums:
        prefix += num
        count += freq.get(prefix - k, 0)
        freq[prefix] = freq.get(prefix, 0) + 1
    return count
````

## 5. Binary Search (LC 704)

````python
# SOLUTION
def binary_search(nums: list[int], target: int) -> int:
    lo, hi = 0, len(nums) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target: return mid
        elif nums[mid] < target: lo = mid + 1
        else: hi = mid - 1
    return -1
````

## 6. Monotonic Stack — Daily Temperatures (LC 739)

````python
# SOLUTION
def daily_temperatures(temps: list[int]) -> list[int]:
    result = [0] * len(temps)
    stack = []
    for i, t in enumerate(temps):
        while stack and temps[stack[-1]] < t:
            j = stack.pop()
            result[j] = i - j
        stack.append(i)
    return result
````

## 7. Linked List — Reverse (LC 206)

````python
# SOLUTION
def reverse_list(head):
    prev = None
    while head:
        nxt = head.next
        head.next = prev
        prev = head
        head = nxt
    return prev
````

## 9. Tree BFS — Level Order (LC 102)

````python
# SOLUTION
def level_order(root) -> list[list]:
    if not root: return []
    result, q = [], deque([root])
    while q:
        level = []
        for _ in range(len(q)):
            node = q.popleft()
            level.append(node.val)
            if node.left:  q.append(node.left)
            if node.right: q.append(node.right)
        result.append(level)
    return result
````

## 10. Graph DFS — Number of Islands (LC 200)

````python
# SOLUTION
def num_islands(grid: list[list[str]]) -> int:
    rows, cols = len(grid), len(grid[0])
    def dfs(r, c):
        if r < 0 or r >= rows or c < 0 or c >= cols or grid[r][c] != '1': return
        grid[r][c] = '0'
        dfs(r+1,c); dfs(r-1,c); dfs(r,c+1); dfs(r,c-1)
    count = 0
    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == '1': dfs(r, c); count += 1
    return count
````

## 11. Backtracking — Subsets (LC 78)

````python
# SOLUTION
def subsets(nums: list[int]) -> list[list[int]]:
    result = []
    def backtrack(start, path):
        result.append(path[:])
        for i in range(start, len(nums)):
            path.append(nums[i])
            backtrack(i + 1, path)
            path.pop()
    backtrack(0, [])
    return result
````

## 12. Heap — Kth Largest (LC 215)

````python
# SOLUTION
def kth_largest(nums: list[int], k: int) -> int:
    heap = []
    for num in nums:
        heapq.heappush(heap, num)
        if len(heap) > k: heapq.heappop(heap)
    return heap[0]
````

## 13. Intervals — Merge (LC 56)

````python
# SOLUTION
def merge(intervals: list[list[int]]) -> list[list[int]]:
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]
    for start, end in intervals[1:]:
        if start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])
    return merged
````

## 14. DP — Climbing Stairs (LC 70)

````python
# SOLUTION
def climb_stairs(n: int) -> int:
    if n <= 2: return n
    a, b = 1, 2
    for _ in range(3, n + 1):
        a, b = b, a + b
    return b
````
